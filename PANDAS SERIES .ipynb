# Accessing elements of a Pandas Series 

```python
import numpy as np
import pandas as pd

elon_ai_comp=np.array(["Tesla","Starlink","X","Solarcity","xAI"])
ser=pd.Series(elon_ai_comp , index=["a","b","c","d","e"])
ser
```

```python
# indexing single value
ser["a"]
```

```python
#indexing a series

ser["a":"d"]
```

```python
# accessing 
ser["a":"e":2]
```

```python
ser[["a","c","e"]]
```

# operation on pandas series 

```python
# adding two series 

s1=pd.Series([10,20,30],index=['a','b','c'])
s2=pd.Series([1,2,3],index=['a','b','c'])

res=s1+s2
print(res)
```

```python
# adding two series 

s1=pd.Series([10,20,30],index=['a','b','c'])
s2=pd.Series([1,2,3],index=['d','e','f'])

res=s1+s2
print(res)
```

```python
# adding two series 

s1=pd.Series([10,20,30],index=['a','b','c'])
s2=pd.Series([1,2,3,4,5,6],index=['a','b','c','d','e','f'])

res=s1+s2
print(res)
```

```python
s1=pd.Series([10,20,30])
s2=pd.Series([10,25,30])

res=s1==s2
print(res)
```

```python
s1[s1==s2]
```

```python
s1[s1!=s2]
```

```python
s1[(s1>5) & (s2>20)]
```

```python
s1=pd.Series([10,20,30])
s2=pd.Series([10,25,35])

res=s1==s2
print(res)
```

```python
s1[(s1>5) & (s2>20)]
```

```python
s1[(s1>20) & (s2>20)]
```

# Pandas series index attribute 

```python
data = pd.Series([10,20,30,40] , index=['a','b','c','d'])

# Accessing the index 
print("origional index",data.index)
```

```python
# modifying the index 

data.index=['w','x','y','z']
print("modified series:\n", data)
```
