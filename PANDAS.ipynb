# MERGING DATAFRAME

  in pandas is similar to performing sql joins ( right outer , left outer ) ,merge() function provide flexibility for different type of joins

  

### **MERGING DATAFRAME USING 1-KEY**  (ON - KEY)

```python
import pandas as pd

DATA_1 = {'key': ['K0','K1','K2','K3'],
         'Name':['jaideep','prince','golu','anuj'],
         'Age':[27,24,22,32]}
         

DATA_2 = {'key': ['K0','K1','K2','K3'],
         'Adress':['Nagpur','Kanpur','Allahabad','Kannauj'],
         'Qualification':['Btech','B.A','Bcom','B.hons']}

df_DATA_1 = pd.DataFrame(DATA_1)

df_DATA_2 = pd.DataFrame(DATA_2)

print(df_DATA_1,"\n\n",df_DATA_2)
```

```python
# NOW HERE WE ARE USING .MERGE() KEY WITH ONE UNIQUE COMBINATION
res = pd.merge(df_DATA_1 , df_DATA_2 , on='key')
res
```

# Group by in pandas followa the split-apply-combine paradigm popularized by handley wickham.

**the groupby() operation divided into three main steps**y

   ***Split-apply-combine(groupb) in python (pandas)***




 

```python
import pandas as pd 

df = pd.DataFrame({
    "Student":["A","B","A","B","A"],
    "Subject":["Math","Math","Science","Science","Math"],
    "Marks":[80,70,85,75,90]
})


df
```

# **Aggregation (reduce rows)**

```python
df.groupby("Student").groups
```

```python
df.groupby("Student").sum()
```

```python
df.groupby("Student")["Marks"].sum()
```

```python
df.groupby("Student")["Subject"].sum()
```

# ***Transform (same shape as origional)***

```python
df["std"] = df.groupby("Student")["Marks"].transform("std")
df
```
